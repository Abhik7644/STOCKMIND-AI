```python
import pandas as pd
import matplotlib.pyplot as plt
import yfinance as yf
```

```python
ticker = "AAPL"
df=yf.download(ticker,start="2019-01-01",end="2026-01-01")
print(df.shape)
print(df.head())
```

```python
print(df.isnull().sum())
print(f"\nDate range: {df.index[0]} -> {df.index[-1]}")
print(f"Total trading days: {len(df)}")
```

```python
plt.figure(figsize=(14, 5))
plt.plot(df['Close'], color='royalblue', linewidth=1.5)
plt.title(f'{ticker} Closing Price (2019–2024)', fontsize=14)
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
df.to_csv('../data/AAPL_raw.csv')
print("Saved!")
```
